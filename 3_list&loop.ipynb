## 2. List Index (Accessing Elements)

Each item in a list has a position number called an **index**. Python starts counting positions from `0`.

```text

```

### Example:
```python

```

## 3. Negative Indexing

Negative indexing means starting from the end of the list. `-1` refers to the last item, `-2` refers to the second-last item, and so on.

```text
List:            [10, 20, 30, 40, 50]
Negative Index:  -5  -4  -3  -2  -1
```

### Example:
```python
names = ["Amit", "Priya", "Vijay"]
print(names[-1]) # Prints "Vijay"
```

## 8. Removing Elements with `pop()`

## 9. Finding Length with `len()` and Checking with `in`

* `len(list)` returns the total number of items in a list.

## 11. Slicing a List

Slicing allows you to get a part of a list using `list[start:stop]`. Note that the `stop` index is **not** included.

### Example:
```python

print(numbers[1:4]) # Gets items from index 1 to 3: [20, 30, 40]
```

## 12. `sort()`, `reverse()`, and `clear()`

* `sort()` arranges a list in ascending order.
* `reverse()` reverses the order of the elements in-place.
* `clear()` deletes all elements from the list, leaving it empty.

### Example:
```python
vals = [3, 1, 2]
vals.sort()
print(vals) # [1, 2, 3]
```
